```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('titanic')
df.head()
```

## 1. Categorical Data
### a. Countplot

```python
sns.countplot(df['embarked'])
```

```python
df['survived'].value_counts().plot(kind='bar')
```

### b. Pie Chart

```python
df['sex'].value_counts().plot(kind='pie', autopct='%0.2f')
```

## 2. Numerical Data
### a. Histogram

```python
plt.hist(df['age'], bins=5)
```

### b. Displot

```python
sns.distplot(df['age'])
```

c. Boxplot

```python
sns.boxplot(df['age'])
```
