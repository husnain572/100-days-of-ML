# ColumnTransformer in Machine Learning (scikit-learn)
Complete theory and practical guide.

## 1. What is ColumnTransformer?
ColumnTransformer is a utility in scikit-learn that allows you to apply different preprocessing steps to different columns of a dataset.

It is especially useful when your dataset contains a mix of numerical and categorical features.

## 2. Why do we need it?
In real-world datasets:
- Numerical features require scaling
- Categorical features require encoding

ColumnTransformer allows handling all transformations in one structured pipeline.

## 3. Syntax
ColumnTransformer([
    ('name1', transformer1, columns1),
    ('name2', transformer2, columns2)
])

## 4. Key Concepts
- Each transformer works on specified columns
- Transformations run in parallel
- Output is combined into a single feature matrix
- Works seamlessly with Pipeline

## 5. Practical Example

```
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```
# Sample dataset
df = pd.DataFrame({
    'Age': [22, 38, 26],
    'Fare': [7.25, 71.83, 8.05],
    'Sex': ['male', 'female', 'female']
})

df
```

```
# Define columns
num_cols = ['Age', 'Fare']
cat_cols = ['Sex']
```

```
# Create ColumnTransformer
preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(), cat_cols)
])
```

```
# Apply transformation
transformed_data = preprocessor.fit_transform(df)
transformed_data
```

## 6. Using with Pipeline (Industry Standard)

```
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ('preprocessing', preprocessor),
    ('model', LogisticRegression())
])
```
