Working with SQL

```python
import sqlite3
conn=sqlite3.connect('students.db')
cursor=conn.cursor()

with open('students.sql', 'r') as f:
  sql_script=f.read()

cursor.executescript(sql_script)

conn.commit()
```

```python
df=pd.read_sql_query('SELECT * FROM students', conn)
```
