# Complete Case Analysis (CCA)

## Definition
Complete Case Analysis (CCA) is a technique used for handling missing data in a dataset. In this method, only those rows are retained where all feature values are present. Any row that contains at least one missing value (NaN) is completely removed from the dataset.

In simple terms, CCA means analyzing only the complete records and ignoring all incomplete ones.

## Example

Consider the following dataset:

| Age | Salary | City       |
|-----|--------|------------|
| 25  | 50000  | Lahore     |
| NaN | 60000  | Karachi    |
| 30  | NaN    | Islamabad  |
| 28  | 55000  | Lahore     |

After applying Complete Case Analysis:

| Age | Salary | City   |
|-----|--------|--------|
| 25  | 50000  | Lahore |
| 28  | 55000  | Lahore |

All rows containing missing values are removed, and only complete records remain.

## Implementation in Python

```python
df.dropna()
````

This function removes all rows that contain at least one missing value.

## When to Use Complete Case Analysis

CCA is appropriate under the following conditions:

* The proportion of missing data is very small (typically less than 5%)
* Data is Missing Completely At Random (MCAR)
* The dataset is large enough so that removing some rows does not significantly impact analysis

## Advantages

* Simple and easy to implement
* Computationally efficient
* No need for estimation or imputation of missing values
* Maintains data integrity for complete records

## Disadvantages

* Loss of valuable data due to row deletion
* Can introduce bias if data is not missing completely at random
* Not suitable for small datasets
* May reduce statistical power due to reduced sample size

## Key Idea

Complete Case Analysis follows a straightforward principle:

Only analyze fully complete records and discard incomplete ones.

`````python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots()

sns.histplot(
    df['training_hours'],
    bins=50,
    ax=ax,
    stat='density',
    color='red',
    alpha=0.5,
    label='Original Data',

)

sns.histplot(
    new_df['training_hours'],
    bins=50,
    ax=ax,
    stat='density',
    color='green',
    alpha=0.5,
    label='After CCA',
      edgecolor='black'

)

ax.legend()
plt.show()
`````

`````python
fig = plt.figure()
ax = fig.add_subplot(111)

# original data
df['training_hours'].plot.density(color='red')

# data after cca
new_df['training_hours'].plot.density(color='green')
`````
