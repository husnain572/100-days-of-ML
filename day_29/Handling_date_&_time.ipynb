```python
# Converting to datetime datatype
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
# extract year
date['date_year']=date['date'].dt.year
```

```python
# extract month
date['date_month']=date['date'].dt.month
```

```python
# extract date
date['date_month_name']=date['date'].dt.month_name
```

```python
# extract day
date['date_day']=date['date'].dt.day
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# Extract week of the year
date['date_week'] = date['date'].dt.isocalendar().week
```

```python
date.sample(4)
```

```python
# Extract Quarter
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# Extract Semester
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# If quarter is 1, 2, or 3 → semester = 1
# If quarter is 4 → semester = 2
date['semester']=np.where(date['quarter'].isin([1,2,3]), 1, 2)
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# gets the current date and time of your system
import datetime
today=datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed

date['months_passed'] = (
    (today.year - date['date'].dt.year) * 12 +
    (today.month - date['date'].dt.month)
).clip(lower=0)
```
